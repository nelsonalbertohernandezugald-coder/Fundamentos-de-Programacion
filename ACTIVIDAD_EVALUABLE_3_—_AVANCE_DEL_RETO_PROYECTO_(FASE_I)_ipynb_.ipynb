```python
import random
from datetime import datetime

def mostrar_catalogo():
    print("\n--- CATÁLOGO DE PRENDAS ---")
    print("Prendas Superiores:")
    print("  1. Playera             $30.00")
    print("  2. Camisa / Blusa      $35.00")
    print("  3. Suéter             $40.00")
    print("  4. Saco / Blazer       $60.00")
    print("  5. Chamarra            $70.00")
    print("Prendas Inferiores:")
    print("  6. Pantalón Mezclilla  $40.00")
    print("  7. Pantalón de Vestir  $45.00")
    print("  8. Short / Berbuda     $30.00")
    print("  9. Falda               $35.00")
    print(" 10. Vestido             $65.00")
    print("---------------------------")

def obtener_prenda_info(opcion):
    prendas = {
        1: ("Playera", 30.0),
        2: ("Camisa / Blusa", 35.0),
        3: ("Suéter", 40.0),
        4: ("Saco / Blazer", 60.0),
        5: ("Chamarra", 70.0),
        6: ("Pantalón Mezclilla", 40.0),
        7: ("Pantalón de Vestir", 45.0),
        8: ("Short / Berbuda", 30.0),
        9: ("Falda", 35.0),
        10: ("Vestido", 65.0)
    }
    return prendas.get(opcion, (None, 0.0))

def main():
    print("===========================================")
    print(" BIENVENIDO A LA TINTORERÍA Y PLANCHADURÍA ")
    print("===========================================")

    nombre = input("Por favor, ingresa tu nombre completo: ")

    while True:
        try:
            prendas = int(input("¿Cuántas prendas deseas ingresar?: "))
            if prendas > 0:
                break
            print("Por favor, ingresa una cantidad mayor a 0.")
        except ValueError:
            print("Entrada no válida. Por favor, ingresa un número entero.")

    pedido = []

    for i in range(1, prendas + 1):
        print(f"\n--- Configurando prenda {i} de {prendas} ---")
        mostrar_catalogo()

        # Selección de tipo de prenda
        while True:
            try:
                num_prendas = int(input(f"Selecciona el número de la prenda {i}: "))
                nombre_prenda, precio_base = obtener_prenda_info(num_prendas)
                if nombre_prenda:
                    break
                print("Opción inválida. Elige un número del 1 al 10.")
            except ValueError:
                print("Por favor, introduce un número válido.")

        # Tratamiento especial
        print("\n¿Esta prenda requiere un tratamiento especial?")
        print("1. Ropa blanca (Lavado óptico) (+$10.00)")
        print("2. Cuidado delicado / Tintorería (+$20.00)")
        print("3. Ninguno (Lavado estándar)")

        cargo_adicional = 0.0
        tratamiento = "Estándar"

        while True:
            try:
                opc_tratamiento = int(input("Selecciona una opción (1-3): "))
                if opc_tratamiento == 1:
                    cargo_adicional = 10.0
                    tratamiento = "Blanca (Especial)"
                    break
                elif opc_tratamiento == 2:
                    cargo_adicional = 20.0
                    tratamiento = "Cuidado Delicado"
                    break
                elif opc_tratamiento == 3:
                    break
                print("Opción inválida. Elige 1, 2 o 3.")
            except ValueError:
                print("Por favor, introduce un número válido.")

        precio_final_prenda = precio_base + cargo_adicional
        pedido.append({
            "nombre": nombre_prenda,
            "tratamiento": tratamiento,
            "precio_unitario": precio_final_prenda
        })

    # Selección de descuento
    print("\n--- SELECCIÓN DE DESCUENTO ---")
    print("1. Profesor (15% de descuento)")
    print("2. Universitario (10% de descuento)")
    print("3. Cliente Frecuente (20% de descuento)")
    print("4. Ninguno")

    descuento_porcentaje = 0.0
    tipo_descuento = "Ninguno"

    while True:
        try:
            opc_desc = int(input("Selecciona tu tipo de perfil para descuento (1-4): "))
            if opc_desc == 1:
                descuento_porcentaje = 0.15
                tipo_descuento = "Profesor (15%)"
                break
            elif opc_desc == 2:
                descuento_porcentaje = 0.10
                tipo_descuento = "Universitario (10%)"
                break
            elif opc_desc == 3:
                descuento_porcentaje = 0.20
                tipo_descuento = "Cliente Frecuente (20%)"
                break
            elif opc_desc == 4:
                break
            print("Opción inválida. Elige un número del 1 al 4.")
        except ValueError:
            print("Por favor, introduce un número válido.")

    # Confirmación del pedido
    while True:
        print("\n=========================================")
        print("       CONFIRMACIÓN DE TU PEDIDO")
        print("=========================================")
        subtotal = 0.0
        for idx, item in enumerate(pedido, 1):
            print(f"{idx}. {item['nombre']} ({item['tratamiento']}) - ${item['precio_unitario']:.2f}")
            subtotal += item['precio_unitario']

        monto_descuento = subtotal * descuento_porcentaje
        total_final = subtotal - monto_descuento

        print("-----------------------------------------")
        print(f"Subtotal: ${subtotal:.2f}")
        print(f"Descuento ({tipo_descuento}): -${monto_descuento:.2f}")
        print(f"Total Estimado: ${total_final:.2f}")
        print("-----------------------------------------")

        confirmar = input("¿Todos los datos son correctos? (S/N): ").strip().upper()
        if confirmar == 'S':
            break
        elif confirmar == 'N':
            print("\nPor favor, vuelve a correr el programa para corregir tus prendas.")
            return
        else:
            print("Opción no válida. Escribe 'S' para Sí o 'N' para No.")

    # Tipo de pago
    print("\n--- MÉTODO DE PAGO ---")
    print("1. Efectivo")
    print("2. Tarjeta de Crédito/Débito")
    print("3. Transferencia Bancaria")

    metodo_pago = ""
    while True:
        try:
            opc_pago = int(input("Selecciona tu método de pago (1-3): "))
            if opc_pago == 1:
                metodo_pago = "Efectivo"
                break
            elif opc_pago == 2:
                metodo_pago = "Tarjeta"
                break
            elif opc_pago == 3:
                metodo_pago = "Transferencia"
                break
            print("Opción inválida. Selecciona 1, 2 o 3.")
        except ValueError:
            print("Por favor, introduce un número válido.")

    # Cálculo del tiempo de entrega
    # Supongamos 2 horas de espera base por prenda
    horas_espera = prendas * 2
    tiempo_entrega = f"{horas_espera} horas"
    if horas_espera >= 24:
        dias = horas_espera // 24
        horas_restantes = horas_espera % 24
        tiempo_entrega = f"{dias} día(s) y {horas_restantes} horas"

    # Impresión del Ticket
    num_cliente = random.randint(1000, 9999)

    print("\n\n=========================================")
    print("           TICKET DE COMPRA")
    print("         TINTORERÍA Y PLANCHADURÍA")
    print("=========================================")
    print(f"Fecha/Hora: {datetime.now().strftime('%d/%m/%Y %H:%M:%S')}")
    print(f"Nro. Cliente: {num_cliente}")
    print(f"Cliente:      {nombre}")
    print("-----------------------------------------")
    print(f"{'Cant. - Prenda (Tratamiento)':<30} | {'Precio':>7}")
    print("-----------------------------------------")
    for item in pedido:
        detalle = f"1x {item['nombre']} ({item['tratamiento']})"
        # Truncar si es muy largo para la alineación del ticket
        if len(detalle) > 29:
            detalle = detalle[:26] + "..."
        print(f"{detalle:<30} | ${item['precio_unitario']:>7.2f}")
    print("-----------------------------------------")
    print(f"Subtotal:                           ${subtotal:>7.2f}")
    print(f"Desc. ({tipo_descuento[:12]}):           -${monto_descuento:>7.2f}")
    print("-----------------------------------------")
    print(f"TOTAL A PAGAR:                      ${total_final:>7.2f}")
    print(f"Método de pago:                     {metodo_pago}")
    print("-----------------------------------------")
    print(f"Tiempo de entrega estimado: {tiempo_entrega}")
    print("=========================================")
    print("      ¡Gracias por su preferencia!")
    print("=========================================")

if __name__ == "__main__":
    main()
```
