```python
# Tabla de Pitágoras
# Nelson Alberto Hernández Ugalde
# 3076797



tabla_pitagoras_11 = []
for i in range(1, 11):
    fila = []
    for j in range(1, 11):
        fila.append(i * j)
    tabla_pitagoras_11.append(fila)

def imprimir_tabla(tabla):
    print("--- TABLA DE PITÁGORAS (10x10) ---")
    for fila in tabla:
        for valor in fila:
            print(valor, end="\t")
        print()
    print("-----------------------------------\n")

imprimir_tabla(tabla_pitagoras_11)

def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]

renglon = int(input("Introduce el renglón (1 al 10): "))
while renglon < 1 or renglon > 11:
    print("Valor fuera de rango. Por favor introduce un número entre 1 y 10.")
    renglon = int(input("Introduce el renglón (1 al 10): "))

columna = int(input("Introduce la columna (1 al 10): "))
while columna < 1 or columna > 11:
    print("Valor fuera de rango. Por favor introduce un número entre 1 y 10.")
    columna = int(input("Introduce la columna (1 al 10): "))

resultado = consultar_producto(tabla_pitagoras_11, renglon, columna)
print(f"\nResultado de la consulta: {renglon} x {columna} = {resultado}")
```
