```python
def seccion_tuplas():
    print("\n--- Parte A: Tuplas ---")
    numeros = (10, 25, 4, 89, 50)
    print(f"El tercer elemento de la tupla original es: {numeros[2]}")

    try:
        num1 = float(input("Ingresa el primer número: "))
        num2 = float(input("Ingresa el segundo número: "))
    except ValueError:
        print("Entrada no válida. Se usarán valores por defecto (0, 0).")
        num1, num2 = 0, 0

    nueva_tupla = numeros + (num1, num2)
    print(f"Nueva tupla creada: {nueva_tupla}")

    lista_ordenada = sorted(list(nueva_tupla))
    print(f"Nueva tupla convertida en lista y ordenada: {lista_ordenada}")

    def sumar_tupla(t):
        total = 0
        for elemento in t:
            total += elemento
        return total

    suma_total = sumar_tupla(nueva_tupla)
    print(f"La suma de todos los elementos de la nueva tupla es: {suma_total}")


def seccion_diccionarios():
    print("\n--- Parte B: Diccionarios ---")
    contactos = {"Ana": "555-1234","Carlos": "555-5678","Sofía": "555-9012"}

    nuevo_nombre = input("Ingresa el nombre del nuevo contacto: ").strip()
    nuevo_tel = input("Ingresa el teléfono: ").strip()
    if nuevo_nombre:
        contactos[nuevo_nombre] = nuevo_tel

    print("\nLista de nombres en contactos:")
    for nombre in contactos.keys():
        print(f"- {nombre}")

    def buscar_telefono(dicc, nombre_buscar):
        return dicc.get(nombre_buscar, None)

    nombre_buscar = input("\nIngresa el nombre del contacto que deseas buscar: ").strip()
    telefono = buscar_telefono(contactos, nombre_buscar)

    if telefono:
        print(f"El teléfono de {nombre_buscar} es: {telefono}")
    else:
        print(f"{nombre_buscar} no está en tus contactos.")


def seccion_excepciones():
    print("\n--- Parte C: Excepciones ---")
    try:
        num1_str = input("Ingresa el primer número entero: ")
        num2_str = input("Ingresa el segundo número entero: ")

        n1 = int(num1_str)
        n2 = int(num2_str)

        suma = n1 + n2
        division = n1 / n2

        print(f"Suma: {suma}")
        print(f"División: {division}")

    except ValueError:
        print("Error: Debes ingresar números enteros válidos (no se permiten letras ni campos vacíos).")
    except ZeroDivisionError:
        print("Error: No es posible dividir entre cero. Por favor ingresa un divisor diferente.")


def seccion_strings():
    print("\n--- Parte D: Strings ---")
    mensaje = "la música conecta a las personas"
    print(f"Mensaje original: \"{mensaje}\"")

    print(f"Longitud del mensaje: {len(mensaje)} caracteres")

    print(f"En mayúsculas: {mensaje.upper()}")

    palabra_buscar = "música"
    palabra_nueva = "tecnología"
    nuevo_mensaje = mensaje.replace(palabra_buscar, palabra_nueva)
    print(f"Mensaje modificado: \"{nuevo_mensaje}\"")

    def contar_palabras(texto):
        palabras = texto.split()
        return len(palabras)

    total_palabras = contar_palabras(mensaje)
    print(f"El mensaje tiene: {total_palabras} palabras")


def ejecutar_menu():
    while True:
        print("\n=========================================")
        print("   CONSOLA DE SONIDO - MENÚ PRINCIPAL")
        print("=========================================")
        print("1) Tuplas")
        print("2) Diccionarios")
        print("3) Excepciones")
        print("4) Strings")
        print("5) Salir")

        opcion = input("Elige una opción: ").strip()

        if opcion == "1":
            seccion_tuplas()
        elif opcion == "2":
            seccion_diccionarios()
        elif opcion == "3":
            seccion_excepciones()
        elif opcion == "4":
            seccion_strings()
        elif opcion == "5":
            print("\n¡Gracias por usar la consola de sonido! Saliendo del programa...")
            break
        else:
            print("\nOpción inválida. Intenta de nuevo escribiendo un número del 1 al 5.")

ejecutar_menu()
```
