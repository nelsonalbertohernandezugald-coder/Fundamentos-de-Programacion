## Paso 1 — Ejecuta la calculadora (versión de 3 plataformas)

Ejecuta la celda de abajo con `Shift + Enter`. El programa te hará preguntas en un cuadro de texto: escribe tu respuesta y presiona `Enter` después de cada una. Usa **tus horas reales**. Para escribir media hora usa punto, no coma: `1.5`.

```python
nombre = input("¿Cómo te llamas? ")
horas_redes = float(input("Horas al día en redes sociales: "))
horas_mensajeria = float(input("Horas al día en mensajería: "))
horas_streaming = float(input("Horas al día en streaming (series, videos): "))

tiempo_total = horas_redes + horas_mensajeria + horas_streaming
porcentaje = (tiempo_total / 24) * 100

print("----- RESUMEN DE TU RUTINA DIGITAL -----")
print(f"Usuario: {nombre}")
print(f"Tiempo total en pantalla: {tiempo_total} horas")
print(f"Porcentaje de tu día: {porcentaje:.2f}%")
```

```python
# Pega aquí el código del Paso 1 y amplíalo a 5 plataformas
nombre = input("¿Cómo te llamas? ")

# Plataforma 1
red1 = input("Nombre de la primera aplicación: ")
horas1 = float(input(f"Horas al día en {red1}: "))

# Plataforma 2
red2 = input("Nombre de la segunda aplicación: ")
horas2 = float(input(f"Horas al día en {red2}: "))

# Plataforma 3
red3 = input("Nombre de la tercera aplicación: ")
horas3 = float(input(f"Horas al día en {red3}: "))

# Plataforma 4
red4 = input("Nombre de la cuarta aplicación: ")
horas4 = float(input(f"Horas al día en {red4}: "))

# Plataforma 5
red5 = input("Nombre de la quinta aplicación: ")
horas5 = float(input(f"Horas al día en {red5}: "))

# Sumamos únicamente las variables numéricas de las horas
tiempo_total = horas1 + horas2 + horas3 + horas4 + horas5
porcentaje = (tiempo_total / 24) * 100

print("----- RESUMEN DE TU RUTINA DIGITAL -----")
print(f"Usuario: {nombre}")
print(f"Tiempo total en pantalla: {tiempo_total} horas")
print(f"Porcentaje de tu día: {porcentaje:.2f}%")
```
