```python
# Personas que ingresaran.
print("Hola muy buen dia, cuantos integrantes pasaran: ")
print("Solo agrege numeros")

while True:
  try:
    integrante = int(input("Integrantes: "))
    if integrante > 0:
      break
    else:
      print("Por favor, ingrese un número mayor a 0.")
  except ValueError:
    print("Entrada no válida. Por favor, introduzca únicamente números enteros.")

if integrante <= 1:
  print("Introduca su edad y el tipo de visitant que es.")
else:
  print("Introsuca las edades y el tipo de visitantes que son.")

# Edades y Tipo de visitante.
print("Los menores de 3 años su entrada es gratuita.")

# Inicializamos el total de ingresos acumulados
total_ingresos = 0.0

# El bucle se ejecutará exactamente el número de veces indicado en 'integrante'
i = 1
while i <= integrante:
  print(f"\n--- Registrando al integrante {i} de {integrante} ---")
  try:
    edad = int(input("Escribe tu edad: "))
  except ValueError:
    print("Error: La edad debe ser un número entero. Inténtelo de nuevo para este integrante.")
    continue

  if edad >= 110 or edad < 0:
    print("Edad no valida")
    # En caso de edad inválida, este integrante no suma costo
    precio_final = 0
  else:
    # Aplicar automáticamente el tipo de visitante según la edad para evitar preguntas innecesarias
    if edad <= 3:
      print("Se detectó automáticamente: Menor de 3 años (Gratis).")
      tipo = 4
    elif edad >= 4 and edad <= 15:
      print("Se detectó automáticamente: Menor de edad (4-15 años) con 10% de descuento.")
      tipo = 5  # Tipo especial para descuento automático de menor
    elif edad >= 16 and edad <= 17:
      print("Se detectó automáticamente: Menor de edad (16-17 años).")
      tipo = 4
    elif edad >= 60 and edad <= 100:
      print("Se detectó automáticamente: Adulto mayor (60-100 años).")
      tipo = 3
    else:
      # Solo se pregunta a personas de entre 18 y 59 años (se eliminó opción 'Adulto mayor')
      while True:
        print("¿Qué tipo de visitante eres?")
        print("1. Estudiante")
        print("2. Profesor")
        print("3. Ninguno")
        print()
        try:
          opcion = int(input("Introduce un valor del 1 al 3: "))
          if opcion in [1, 2, 3]:
            if opcion == 1:
              tipo = 1  # Estudiante
            elif opcion == 2:
              tipo = 2  # Profesor
            elif opcion == 3:
              tipo = 4  # Ninguno
            break
          else:
            print("Opción inválida. Debe ser un número del 1 al 3.\n")
        except ValueError:
          print("Entrada no válida. Por favor, introduzca un número del 1 al 3.\n")

    # Determinar el precio base según la edad y el tipo de visitante
    precio = 0
    if edad <= 3:
      precio = 0
    elif tipo == 1 or tipo == 2:
      # Estudiantes y profesores tienen un precio base especial de $30
      precio = 30
    elif (edad >= 4) and (edad <= 17):
      precio = 30
    elif (edad >= 18):
      precio = 45

    # Calcular descuentos aplicables según el tipo de visitante
    if tipo == 1:
      descuento = precio * 0.1
      precio_final = precio - descuento
      print("¡Descuento de estudiante aplicado (10%)!")
      print(f"El pago final es de ${precio_final:.2f}")
    elif tipo == 2:
      descuento = precio * 0.1
      precio_final = precio - descuento
      print("¡Descuento de profesor aplicado (10%)!")
      print(f"El pago final es de ${precio_final:.2f}")
    elif tipo == 3:
      descuento = precio * 0.12
      precio_final = precio - descuento
      print("¡Descuento de adulto mayor aplicado (12%) automáticamente!")
      print(f"El pago final es de ${precio_final:.2f}")
    elif tipo == 5:
      descuento = precio * 0.1
      precio_final = precio - descuento
      print("¡Descuento automático para menores de 4 a 15 años aplicado (10%)!")
      print(f"El pago final es de ${precio_final:.2f}")
    else:
      precio_final = precio
      if precio_final == 0:
        print("No se cobra.")
      else:
        print(f"El precio base es de ${precio_final:.2f}.")

    # Sumar el costo de este visitante al total acumulado
    total_ingresos += precio_final

  # Avanzar al siguiente integrante sólo si la edad fue procesada
  i += 1

print("\nRegistro finalizado. ¡Hasta luego!")
print(f"El total de ingresos por todos los visitantes es de: ${total_ingresos:.2f}")
```
