```python
# BLOQUE 1 · La rodada (tuplas + listas + ciclos + funciones)
# Ejercicio 1 — Bitácora de la semana

dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(lista_km):
  total_km = 0
  dias_rodados = 0
  max_km = 0
  max_dia = 0

  for i in range(len(lista_km)):
    if lista_km[i] != 0:
      dias_rodados += 1
      total_km += lista_km[i]
      if lista_km[i] > max_km:
        max_km = lista_km[i]
        max_dia = i

  promedio = round(total_km / dias_rodados, 1) if dias_rodados > 0 else 0.0
  return total_km, dias_rodados, promedio, max_dia

total_km, dias_rodados, promedio, max_dia = resumen(km)

print(f"Total: {total_km} km")
print(f"Días con rodada: {dias_rodados}")
print(f"Promedio por día rodado: {promedio} km")
print(f"Día más largo: {dias[max_dia]} ({km[max_dia]} km)")
```

```python
# Ejercicio 2 — La votación del festival

votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

ganador_banda = 0
ganador_votos = 0

for i in range(len(votos)):
  if votos[i][1] > ganador_votos:
    ganador_banda = votos[i][0]
    ganador_votos = votos[i][1]

def promedio_votos(lista_votos):
  total_votos = 0
  for i in range(len(lista_votos)):
    total_votos += lista_votos[i][1]
  return round(total_votos / len(lista_votos), 1)

promedio = promedio_votos(votos)

print(f"Ganador: {ganador_banda} con {ganador_votos} votos")
print(f"Promedio: {promedio:.1f}")
print("Arriba del promedio:")
for banda, cant_votos in votos:
  if cant_votos > promedio:
    print(f"- {banda} ({cant_votos} votos)")
```

```python
# Ejercicio 3 — Tramos de la ruta a la Sierra

paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

max_distancia = 0
tramo_largo = 0

for i in range(len(paradas) - 1):
  inicio, km_inicio = paradas[i]
  destino, km_destino = paradas[i + 1]
  distancia = km_destino - km_inicio
  print(f"{inicio} → {destino}: {distancia} km")

  if distancia > max_distancia:
    max_distancia = distancia
    tramo_largo = f"{inicio} -> {destino} ({distancia} km)"

print(f"Tramo más largo: {tramo_largo}")
```

```python
# Ejercicio 4 — ¿Con chamarra o sin chamarra?

temperaturas = (8, 14, 22, 27, 31, 19, 11)

def calificar(t):
  if t < 12:
    return "frío"
  elif t <= 24:
    return "templado"
  else:
    return "calor"

etiquetas = []
for temperatura in temperaturas:
  etiquetas.append(calificar(temperatura))

contador = {}
for etiqueta in etiquetas:
  if etiqueta in contador:
    contador[etiqueta] += 1
  else:
    contador[etiqueta] = 1

predomina = 0
max_cuenta = 0
for clima, cuenta in contador.items():
  if cuenta > max_cuenta:
    max_cuenta = cuenta
    predomina = clima

print("Etiquetas:", etiquetas)
print("Frecuencias:", contador)
print(f"Predomina: {predomina}")
```

```python
# Ejercicio 5 — ¿Cuánto dura el show?

setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    if segundos < 10:
        return f"{minutos}:0{segundos}"
    else:
        return f"{minutos}:{segundos}"

total_segundos = 0
for cancion, duracion in setlist:
    total_segundos += duracion
    print(f"{cancion}: {formato(duracion)}")

print(f"Duración total: {formato(total_segundos)}")
```

```python
# BLOQUE 2 · El taller y la tienda (diccionarios + listas + funciones)
# Ejercicio 6 — Pedidos del taller

inventario = {"bujía": 10, "cadena": 2, "balata": 6, "filtro": 4}
pedidos = [("bujía", 3), ("cadena", 5), ("espejo", 1), ("balata", 6), ("filtro", 1)]

for pieza, cantidad in pedidos:
    if pieza not in inventario:
        print(f"{pieza}: no existe en el taller")
    elif inventario[pieza] < cantidad:
        print(f"{pieza}: stock insuficiente (hay {inventario[pieza]})")
    else:
        inventario[pieza] -= cantidad
        print(f"{pieza}: surtido ({cantidad})")

agotados = []
for pieza, stock in inventario.items():
    if stock == 0:
        agotados.append(pieza)

print(inventario)
print(f"Agotados: {agotados}")
```

```python
# Ejercicio 7 — Boleta del grupo

notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(lista):
    suma = 0
    for nota in lista:
        suma += nota
    return round(suma / len(lista), 1)

promedios_alumnos = {}
mejor_promedio = 0
mejor_alumno = 0

for alumno, calificaciones in notas.items():
    prom = promedio(calificaciones)
    promedios_alumnos[alumno] = prom

    estado = "Aprobado" if prom >= 70 else "Reprobado"
    print(f"{alumno}: {prom} - {estado}")

    if prom > mejor_promedio:
        mejor_promedio = prom
        mejor_alumno = alumno

print(f"Mejor promedio: {mejor_alumno}")
```

```python
# Ejercicio 8 — Agenda por inicial

nombres = ["Ana", "Alberto", "Beto", "Carla", "Brenda", "Andrea", "Ceci", "Diego"]

grupos = {}

for nombre in nombres:
    inicial = nombre[0]
    if inicial not in grupos:
        grupos[inicial] = []
    grupos[inicial].append(nombre)

for inicial, lista in sorted(grupos.items()):
    print(f"{inicial} ({len(lista)}): {lista}")
```

```python
#Ejercicio 9 — El ticket del merch

precios = {"playera": 350, "gorra": 220, "parche": 80, "sudadera": 650}
carrito = ["playera", "gorra", "playera", "parche", "parche", "parche"]

contador = {}
for articulo in carrito:
    if articulo in contador:
        contador[articulo] += 1
    else:
        contador[articulo] = 1

subtotal = 0
for articulo, cantidad in contador.items():
    precio_unitario = precios[articulo]
    importe = cantidad * precio_unitario
    subtotal += importe
    print(f"{cantidad} x {articulo} = ${importe}")

descuento = 0.0
if subtotal > 800:
    descuento = subtotal * 0.10

total = subtotal - descuento

print(f"Subtotal: ${subtotal}")
print(f"Descuento: ${descuento:.1f}")
print(f"Total: ${total:.1f}")
```

```python
# Ejercicio 10 — Traductor de mensajes de WhatsApp

abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    palabras_expandidas = []
    for palabra in palabras:
        palabras_expandidas.append(abrev.get(palabra, palabra))
    return " ".join(palabras_expandidas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

```python
# BLOQUE 3 · A prueba de usuarios (excepciones + validación)
# Ejercicio 11 — Rendimiento de la moto

def rendimiento():
    while True:
        try:
            entrada_km = input("KM recorridos: ")
            km = float(entrada_km)

            entrada_litros = input("Litros cargados: ")
            litros = float(entrada_litros)

            if litros == 0:
                raise ZeroDivisionError

            r_calculado = round(km / litros, 1)

            if r_calculado >= 20:
                clasificacion = "Excelente"
            elif r_calculado >= 15:
                clasificacion = "Normal"
            else:
                clasificacion = "Revisa tu moto"

            return r_calculado, clasificacion

        except ValueError:
            print("Escribe solo números")
        except ZeroDivisionError:
            print("Los litros no pueden ser 0")

rend, clase = rendimiento()
print(f"\nRendimiento: {rend} km/l")
print(clase)
```

```python
# Ejercicio 12 — Taquilla familiar

def precio_boleto(edad):
    if edad < 12:
        return 150
    elif edad <= 59:
        return 300
    else:
        return 180

total_pagar = 0

for i in range(1, 4):
    while True:
        try:
            entrada = input(f"Edad de la persona {i}: ")
            edad = int(entrada)

            if edad < 0 or edad > 120:
                print("Edad fuera de rango")
                continue

            costo = precio_boleto(edad)
            print(f"Boleto: ${costo}")
            total_pagar += costo
            break

        except ValueError:
            print("Escribe la edad con número")

print(f"Total a pagar: ${total_pagar}")
```

```python
#Ejercicio 13 — Capturas sucias

capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

validas = []
invalidas = 0

for elemento in capturas:
    try:
        calificacion = int(elemento)
        validas.append(calificacion)
    except ValueError:
        invalidas += 1

promedio = round(sum(validas) / len(validas), 2) if len(validas) > 0 else 0.0

print(f"Válidas: {validas}")
print(f"Inválidas: {invalidas}")
print(f"Promedio: {promedio}")
```

```python
# Ejercicio 14 — Reserva tu asiento

sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]

def reservar(sala):
    while True:
        try:
            fila = int(input("Fila (0-2): "))
            columna = int(input("Columna (0-2): "))

            if sala[fila][columna] == "X":
                print("Ocupado, elige otro")
                continue

            sala[fila][columna] = "X"
            print("¡Reservado!")
            return True

        except ValueError:
            print("Escribe números")
        except IndexError:
            print("Ese asiento no existe")

reservar(sala)

for fila in sala:
    for asiento in fila:
        print(asiento, end=" ")
    print()
```

```python
# Ejercicio 15 — Cupones de descuento

cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def aplicar(total, codigo):
    try:
        codigo_limpio = codigo.strip().upper()
        descuento = cupones[codigo_limpio]
        total_descontado = total * (1 - descuento / 100)
        return total_descontado
    except KeyError:
        print("Cupón no válido")
        return total

total_inicial = 1200
codigos_prueba = ["  moto20 ", "GRATIS", "metal10"]

for cod in codigos_prueba:
    nuevo_total = aplicar(total_inicial, cod)
    if nuevo_total == total_inicial and cod.strip().upper() not in cupones:
        print(f"'{cod}' → ${nuevo_total}")
    else:
        print(f"'{cod}' → ${nuevo_total:.1f}")
```

```python
# BLOQUE 4 · El cartel (strings + todo lo demás)
# Ejercicio 16 — ¿Tu contraseña aguanta?

def evaluar(pw):
    fallas = []

    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")

    tiene_numero = False
    tiene_mayuscula = False
    tiene_espacio = False

    for caracter in pw:
        if caracter.isdigit():
            tiene_numero = True
        if caracter.isupper():
            tiene_mayuscula = True
        if caracter == " ":
            tiene_espacio = True

    if not tiene_numero:
        fallas.append("falta un número")

    if not tiene_mayuscula:
        fallas.append("falta una mayúscula")

    if tiene_espacio:
        fallas.append("sin espacios")

    return fallas if len(fallas) > 0 else "segura"

passwords_prueba = ["metal", "Harley2008", "road glide 08", "KLR650moto"]

for contrasena in passwords_prueba:
    resultado = evaluar(contrasena)
    if resultado == "segura":
        print(f"{contrasena}: segura")
    else:
        print(f"{contrasena}: {resultado}")
```

```python
# Ejercicio 17 —  Se lee igual al revés

def es_palindromo(texto):
    texto_limpio = texto.lower().replace(" ", "")
    return texto_limpio == texto_limpio[::-1]

frases_prueba = ("Anita lava la tina", "ABBA", "Metallica", "Oso")

for frase in frases_prueba:
    resultado = es_palindromo(frase)
    print(f"{frase}: {resultado}")
```
